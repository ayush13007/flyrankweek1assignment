## The Question

**Decision:** Which webpages should be reviewed first for content refresh?

**Who acts on it:** The content or SEO team.

**Unit of Analysis:** One webpage (one URL).

**Output:** A score or prediction showing how likely a page is to need a content refresh.

**Action:** Pages with the highest scores will be prioritized for review and updating.

**Cost of a Wrong Recommendation:**

- **False Positive:** A healthy page is refreshed unnecessarily, wasting time and resources.
- **False Negative:** A page that is actually declining is missed, causing continued traffic loss and missed opportunities.

**Why Machine Learning Helps:**

Machine learning can identify patterns from multiple website performance metrics and help rank pages that deserve attention better than relying on a single rule.

## Quick Look at the Data

I explored the starter dataset to understand whether it is suitable for this project.

Some key observations are:

- The dataset contains **30,000 webpages**.
- **16,262 pages** are labelled with a **"down"** trend, making declining content the largest group.
- The median content age for pages with a **"down"** trend is **216 days**, suggesting that older content may require review, although age alone is not enough to decide.

These observations suggest that the dataset contains meaningful variation and can support a content refresh prioritization project.

```python
import pandas as pd

df = pd.read_csv("data/raw/content_refresh_anonymized.csv")

print("Total pages:", len(df))

print("\nTrend direction counts:")
print(df["trend_direction"].value_counts())

print("\nMedian content age (days):")
print(df.groupby("trend_direction")["content_age_days"].median())
```
